### Calculate market risk premiums using three approaches

```python
mrp = MarketRiskPremium(rfr, ret, inf)

mrp_i = mrp.ibbotson_approach()
print(f"Ibbotson Approach: {mrp_i:.4f}")

mrp_w = mrp.wright_approach(0.028)
print(f"Wright Approach: {mrp_w:.4f}")

mrp_d = mrp.damodaran_approach()
print(f"Damodaran Approach: {mrp_d:.4f}")

avg_mrp = (mrp_i + mrp_w + mrp_d) / 3
print(f"Average MRP: {avg_mrp:.4f}")
```

### Instantiate cost of equity and cost of debt

```python
coe = CostOfEquity(rfr, [mrp_i, mrp_w, mrp_d])
cod = CostOfDebt(cb, tb_mat, tb_yld)
```

### Summary

```python
import pandas as pd

def get_results(
    company: str,
    equity_beta: float,
    asset_beta: float,
    gearing: float,
    de_ratio: float,
    coe: float,
    cod: float,
    wacc_pre_tax: float,
    wacc_vanilla: float,
    wacc_post_tax: float
):
    return {
        "Company": company,
        "Equity Beta": equity_beta,
        "Asset Beta": asset_beta,
        "Gearing Ratio": gearing,
        "Debt to Equity Ratio": de_ratio,
        "Cost of Equity (%)": coe * 100,
        "Cost of Debt (%)": cod * 100,
        "Pre Tax WACC (%)": wacc_pre_tax * 100,
        "Vanilla WACC (%)": wacc_vanilla * 100,
        "Post Tax WACC (%)": wacc_post_tax * 100,
    }

jb_results = get_results(
    pub_company,
    jb_data["Equity Beta"],
    jb_data["Asset Beta"],
    jb_data["Gearing Ratio"],
    jb_data["Debt to Equity Ratio"],
    coe_jb,
    cod_jb,
    wacc_jb.pre_tax(),
    wacc_jb.vanilla(),
    wacc_jb.post_tax()
)

alinta_results = get_results(
    priv_company,
    alinta_equity_beta,
    alinta_asset_beta,
    alinta_gearing,
    alinta_de_ratio,
    coe_alinta,
    cod_alinta,
    wacc_alinta.pre_tax(),
    wacc_alinta.vanilla(),
    wacc_alinta.post_tax()
)

df = pd.DataFrame([jb_results, alinta_results])

display(df)
```
